# Week 36

```python
from pathlib import Path

from ku_nlp_group_25_project.week_36.classifier import Classifier
from ku_nlp_group_25_project.utilities import load_dataset


results_dir = Path.cwd().parent / "results"
results_dir.mkdir(exist_ok=True, parents=True)

training_data, validation_data = load_dataset()
```

## Baseline 2: Deterministic rule-based classification

```python
classifier = Classifier(
    casefolding=True,
    min_shared_tokens=1,
    normalization_form="NFC",
)
predictions = classifier.predictions(
    dataframe=validation_data,
)
metrics = Classifier.evaluate(
    predictions_dataframe=predictions,
)

predictions.to_csv(
    results_dir / "week_36_baseline_2_predictions.csv",
    index=False,
)
metrics.to_csv(
    results_dir / "week_36_baseline_2_metrics.csv",
    index=False,
)
```

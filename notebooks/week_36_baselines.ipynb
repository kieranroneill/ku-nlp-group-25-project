# Week 36: Baselines

```python
from pathlib import Path

from ku_nlp_group_25_project.baselines.classifier import Classifier
from ku_nlp_group_25_project.utilities import load_dataset

outputs_dir = Path.cwd().parent / "outputs" / "week_36"
outputs_dir.mkdir(exist_ok=True, parents=True)

training_data, validation_data = load_dataset()
```

## 2. Baseline 2: Deterministic rule-based classification

```python
classifier = Classifier(
    casefolding=True,
    min_shared_tokens=1,
    normalisation_form="NFC",
)
predictions = classifier.predictions(
    dataframe=validation_data,
)
metrics = classifier.evaluate(
    predictions_dataframe=predictions,
)

predictions.to_csv(
    outputs_dir / "rule_based_classifier_predictions.csv",
    index=False,
)
metrics.to_csv(
    outputs_dir / "rule_based_classifier_metrics.csv",
    index=False,
)
```

# Week 38: Pure Trigram Model Analysis

```python
from time import perf_counter

from ku_nlp_group_25_project.constants import UNK
from ku_nlp_group_25_project.language_models.pure_trigram import PureTrigramLanguageModel
from ku_nlp_group_25_project.utilities import load_dataset


training_data, validation_data = load_dataset()

# get the koeran only questions
training_questions = training_data.loc[
    training_data["lang"] == "ko",
    "question",
].tolist()
validation_questions = validation_data.loc[
    validation_data["lang"] == "ko",
    "question",
].tolist()
```

## 1. Model Training

```python
model = PureTrigramLanguageModel(
    minimum_count=2,
    smoothing=0.1,
)
start = perf_counter()

model.train(
    questions=training_questions,
)

print(f"Training time: {perf_counter() - start:.6f} seconds")
```

## 2. Evaluate Perplexity

```python
summary = model.summary(
    validation_questions=validation_questions,
)
lowest_perplexity = sorted(
    summary.question_scores,
    key=lambda score: (
        score.perplexity,
        score.index,
    ),
)[:3]
highest_perplexity = sorted(
    summary.question_scores,
    key=lambda score: (
        -score.perplexity,
        score.index,
    ),
)[:3]

print(f"Summary")
print("=" * len("Summary"))
print(f"Vocabulary size: {summary.vocabulary_size}")
print(f"Unknown token rate: {summary.unknown_rate:.2%}")
print(f"Perplexity: {summary.perplexity:.4f}")

for heading, scores in [
    ("Lowest perplexities", lowest_perplexity),
    ("Highest perplexities", highest_perplexity),
]:
    print(f"\n{heading}")
    print("=" * len(heading))

    for score in scores:
        print(f"\nQuestion index: {score.index}")
        print(f"Question: {score.question}")
        print(f"Tokens: {' | '.join(score.tokens)}")
        print(
            "Unknown token rate: "
            f"{score.tokens.count(UNK)}/{len(score.tokens)}"
        )
        print(f"Perplexity: {score.perplexity:.4f}")
```
